```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read the data
df = pd.read_csv('Housing.csv', usecols=['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking'])
df.head()
```

```python

# Getting the input and output values
X = df.drop('price', axis=1).values
y = df['price'].values.reshape(-1, 1)

# Normalize things to make the training smoother
X = (X - X.mean(axis=0)) / X.std(axis=0)
y = (y - y.mean()) / y.std()
```

```python
from sklearn.preprocessing import StandardScaler

scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_columns = ['area', 'bedrooms', 'bathrooms', 'stories', 'parking']
y_column = ['price']

df[X_columns] = scaler_X.fit_transform(df[X_columns])
df[y_column] = scaler_y.fit_transform(df[y_column])
```

```python
def initialize_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dims)

    for l in range(1, L):
        parameters['W' + str(l)] = np.ones((layer_dims[l], layer_dims[l - 1])) * 0.1
        parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))

    return parameters
```

```python
def linear_forward(A_prev, W, b):
    return np.dot(W, A_prev) + b
```

```python
def L_layer_forward(X, parameters):
    A = X
    L = len(parameters) // 2  # number of layers

    for l in range(1, L + 1):
        Wl = parameters['W' + str(l)]
        bl = parameters['b' + str(l)]
        A = linear_forward(A, Wl, bl)

    return A, X  # return X for backprop update
```

```python

def update_parameters(parameters, y, y_hat, A1, X, learning_rate=0.001):
    error = y - y_hat  # scalar

    # Output layer updates
    for i in range(5):  # 5 neurons in hidden layer
        parameters['W2'][0][i] += learning_rate * 2 * error * A1[i][0]
    parameters['b2'][0][0] += learning_rate * 2 * error

    # Hidden layer updates (assumes no activation so derivative = 1)
    for i in range(5):  # 5 neurons in hidden layer
        for j in range(5):  # 5 inputs
            parameters['W1'][i][j] += learning_rate * 2 * error * parameters['W2'][0][i] * X[j][0]
        parameters['b1'][i][0] += learning_rate * 2 * error * parameters['W2'][0][i]

```

```python
parameters = initialize_parameters([5, 5, 1])
epochs = 5

for i in range(epochs):
    Loss = []

    for j in range(df.shape[0]):
        X = df[['area', 'bedrooms', 'bathrooms', 'stories', 'parking']].values[j].reshape(5, 1)
        y = df[['price']].values[j][0]

        y_hat, A1 = L_layer_forward(X, parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameters, y, y_hat, A1, X)
        Loss.append((y - y_hat) ** 2)

    print('Epoch -', i + 1, 'Loss -', np.mean(Loss))

```
